```python
from models.shared.artifacts import load_model_bundle, predict_calibrated_quantiles

# Calibrated quantile XGBoost (PIT map stored beside the trees).
MODEL_PATH = ROOT / "models/saved_models/pts_nba_model_production.joblib"

bundle = load_model_bundle(MODEL_PATH)
PTS_FEATURES = list(bundle["feature_names"])
TAUS = np.asarray(bundle["calibration"]["taus"], dtype=float)
KNOT_COLS = [f"q_{t:.2f}" for t in TAUS]

print(MODEL_PATH.name)
print("calibration:", bundle["calibration"]["method"])
print("quantiles:", list(TAUS))
print("n features:", len(PTS_FEATURES))
```

```python
"""
Sample points from the model's predicted quantile function and turn the draws
into P(over) / P(push) / P(under) for a given line.

Inputs per row: 11 predicted knots at TAUS (one value per quantile).
Tails beyond 0.05 / 0.95 are explicit assumptions:
  - lower: linear from 0 (u=0) to the q05 knot
  - upper: exponential, with the scale taken from the q90->q95 slope
"""


def predict_knots(bundle, X):
    """Calibrated knots from the saved PIT bundle. Shape (n_rows, 11)."""
    knots = predict_calibrated_quantiles(bundle, X)
    grid = np.column_stack([knots[c].to_numpy(dtype=float) for c in KNOT_COLS])
    grid = np.maximum(grid, 0.0)
    return np.sort(grid, axis=1)


def sample_from_knots(knots, n_draws=20_000, rng=None, min_tail_scale=0.5):
    """
    knots: (n_rows, 11) sorted predicted quantiles.
    Returns continuous draws, shape (n_rows, n_draws).
    The same uniforms u are shared across rows (common random numbers).
    """
    rng = np.random.default_rng() if rng is None else rng
    knots = np.asarray(knots, dtype=float)
    n = knots.shape[0]
    u = rng.random(n_draws)
    out = np.empty((n, n_draws))

    mid = (u >= TAUS[0]) & (u <= TAUS[-1])
    um = u[mid]
    idx = np.clip(np.searchsorted(TAUS, um, side="right") - 1, 0, len(TAUS) - 2)
    w = (um - TAUS[idx]) / (TAUS[idx + 1] - TAUS[idx])
    out[:, mid] = knots[:, idx] + w * (knots[:, idx + 1] - knots[:, idx])

    lo = u < TAUS[0]
    out[:, lo] = knots[:, [0]] * (u[lo] / TAUS[0])

    hi = u > TAUS[-1]
    b = np.maximum((knots[:, -1] - knots[:, -2]) / np.log(2.0), min_tail_scale)
    out[:, hi] = knots[:, [-1]] + b[:, None] * np.log((1 - TAUS[-1]) / (1 - u[hi]))

    return out


def discretize(draws):
    """Clip at 0 and round half up to integer points."""
    return np.floor(np.clip(draws, 0.0, None) + 0.5)


def prob_over(knots, lines, n_draws=20_000, seed=42, chunk=2_000):
    """
    knots: (n_rows, 11); lines: (n_rows,) e.g. 24.5 or 25.0.
    Returns dict of arrays: p_over, p_push, p_under, and the draw-based mean.
    Half-point lines never push. Whole-number lines can.
    """
    knots = np.asarray(knots, dtype=float)
    lines = np.asarray(lines, dtype=float)
    rng = np.random.default_rng(seed)
    n = knots.shape[0]
    p_over, p_push, p_under, mean = (np.empty(n) for _ in range(4))

    for s in range(0, n, chunk):
        sl = slice(s, s + chunk)
        d = discretize(sample_from_knots(knots[sl], n_draws, rng))
        ln = lines[sl][:, None]
        p_over[sl] = (d > ln).mean(axis=1)
        p_push[sl] = (d == ln).mean(axis=1)
        p_under[sl] = (d < ln).mean(axis=1)
        mean[sl] = d.mean(axis=1)

    return {"p_over": p_over, "p_push": p_push, "p_under": p_under, "mean": mean}


def pseudo_lines(rolling_median, offsets=(-4, -2, 0, 2, 4)):
    """Round rolling median to the nearest .5, then shift. Returns (n, len(offsets))."""
    base = np.floor(np.asarray(rolling_median, float)) + 0.5
    return np.column_stack([np.maximum(base + o, 0.5) for o in offsets])


def log_loss_brier(p, y_over, eps=1e-6):
    p = np.clip(p, eps, 1 - eps)
    ll = -np.mean(y_over * np.log(p) + (1 - y_over) * np.log(1 - p))
    return ll, np.mean((p - y_over) ** 2)


# Self-check against a known negative-binomial quantile function.
from scipy.stats import nbinom

_rng = np.random.default_rng(0)
_mus = _rng.uniform(4, 28, size=500)
_r = 4.0
_knots = np.column_stack([nbinom.ppf(t, _r, _r / (_r + _mus)) for t in TAUS]).astype(float)
_knots = np.sort(_knots, axis=1)
_lines = np.floor(_mus) + 0.5
_res = prob_over(_knots, _lines, n_draws=50_000)
_true = 1 - nbinom.cdf(np.floor(_lines), _r, _r / (_r + _mus))
print("mean abs diff vs true NB P(over):", np.abs(_res["p_over"] - _true).mean().round(4))
print("p_over + p_push + p_under == 1  :", np.allclose(_res["p_over"] + _res["p_push"] + _res["p_under"], 1))
```

```python
from src.live.predict_points import FEATURES, build_points_live, lineups_on, pinnacle_games_on

if list(PTS_FEATURES) != list(FEATURES):
    raise AssertionError("bundle feature_names do not match src.live.predict_points.FEATURES")

games = pinnacle_games_on(day)
lineups = lineups_on(day)
live = build_points_live(
    df,
    players,
    day,
    games=games,
    lineups=lineups,
    extras=("pts_lag_1",),
)
need = ["starting", "pts_lag_1", *FEATURES]
seen, cols = set(), []
for col in need:
    if col not in seen:
        seen.add(col)
        cols.append(col)
X = live.loc[:, cols]

knots_by_player = predict_knots(bundle, X)
pred = pd.DataFrame(knots_by_player, index=X.index, columns=KNOT_COLS)
print(
    f"predicted {len(pred)} players  "
    f"opp/pace missing {int(live[['opp_def_rating_mean_10', 'expected_pace']].isna().any(axis=1).sum())}"
)
pred
```

```python
from models.shared.market import american_to_decimal, no_vig_over

name_to_row = {name: i for i, name in enumerate(X.index)}
rows = markets["player_name"].map(name_to_row)
ok = rows.notna()
knots = knots_by_player[rows[ok].to_numpy(dtype=int)]
lines = markets.loc[ok, "line"].to_numpy(dtype=float)

probs = prob_over(knots, lines, n_draws=20_000, seed=42)
board = markets.loc[ok].copy()
board["q50"] = knots_by_player[rows[ok].to_numpy(dtype=int), KNOT_COLS.index("q_0.50")]
board["mean"] = probs["mean"]
board["p_over"] = probs["p_over"]
board["p_push"] = probs["p_push"]
board["p_under"] = probs["p_under"]

board["no_vig_imp_prob_over"] = no_vig_over(board["over_american"], board["under_american"])
board["no_vig_imp_prob_under"] = 1.0 - board["no_vig_imp_prob_over"]
over_profit = american_to_decimal(board["over_american"]) - 1.0
under_profit = american_to_decimal(board["under_american"]) - 1.0
board["ev_over"] = board["p_over"] * over_profit - board["p_under"]
board["ev_under"] = board["p_under"] * under_profit - board["p_over"]

print(f"slate {day}  players {len(X)}  lines {len(board)}")
print(
    "p_over+p_push+p_under==1:",
    np.allclose(board["p_over"] + board["p_push"] + board["p_under"], 1),
)
board.sort_values(["player_name", "book"]).reset_index(drop=True)
```
